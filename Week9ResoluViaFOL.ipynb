```python
try:
    import nltk
except ImportError:
    !pip install nltk
    import nltk

from nltk.sem import Expression
from nltk.inference import ResolutionProver

# Helper to read FOL formulas from the user
def get_user_input():
    print("--- FOL Resolution Prover ---")
    print("Enter formulas in NLTK FOL format (e.g., 'all x. (man(x) -> mortal(x))', 'man(socrates)')")
    print("Enter empty line to finish entering premises.\n")

    read_expr = Expression.fromstring
    premises = []

    while True:
        user_in = input("Enter a premise: ").strip()
        if not user_in:
            break
        try:
            expr = read_expr(user_in)
            premises.append(expr)
        except Exception as e:
            print(f"Error parsing expression: {e}. Please try again.")

    while True:
        query_in = input("\nEnter the query/goal to prove (e.g., 'mortal(socrates)'): ").strip()
        if query_in:
            try:
                query = read_expr(query_in)
                break
            except Exception as e:
                print(f"Error parsing query: {e}. Please try again.")

    return premises, query

def main():
    # For demo purposes, we will provide default inputs if the user doesn't want to type them
    print("Would you like to enter your own FOL formulas? (y/n)")
    choice = input().strip().lower()

    read_expr = Expression.fromstring
    if choice == 'y':
        premises, query = get_user_input()
    else:
        print("\nUsing default example:")
        print("Premise 1: all x. (man(x) -> mortal(x))")
        print("Premise 2: man(socrates)")
        print("Query: mortal(socrates)\n")
        premises = [read_expr('all x. (man(x) -> mortal(x))'), read_expr('man(socrates)')]
        query = read_expr('mortal(socrates)')

    print("\n--- Running FOL Resolution ---")
    prover = ResolutionProver()
    result = prover.prove(query, premises, verbose=True)

    print(f"\nResult: The query is {'PROVED (True)' if result else 'NOT PROVED (False)'}")

if __name__ == "__main__":
    main()
```
